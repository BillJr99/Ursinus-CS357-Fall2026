```python
import numpy as np
```

```python
# Sample data: x and y
x = np.array([1, 2, 3, 4, 5], dtype=float)  # Input features
y = np.array([3, 5, 7, 9, 11], dtype=float)  # Actual values

# Hyperparameters
learning_rate = 0.01
epochs = 3000
n = len(x)
```

```python
# Initialize weights and biases for the hidden layer and output layer
w1 = np.random.randn()
b1 = np.random.randn()
w2 = np.random.randn()
b2 = np.random.randn()
v1 = np.random.randn()
v2 = np.random.randn()
c = np.random.randn()
```

```python
# ReLU activation function
def relu(z):
    return np.maximum(0, z)

# Derivative of ReLU for backpropagation
def relu_derivative(z):
    return np.where(z > 0, 1, 0)
```

```python
# Training the network
for epoch in range(epochs):
    # Forward pass: compute hidden layer activations
    z1 = w1 * x + b1  # First hidden neuron
    z2 = w2 * x + b2  # Second hidden neuron

    h1 = relu(z1)  # Activation of first hidden neuron
    h2 = relu(z2)  # Activation of second hidden neuron

    # Compute output layer prediction
    y_pred = v1 * h1 + v2 * h2 + c

    # Loss (MSE)
    loss = (1 / n) * np.sum((y - y_pred) ** 2)

    # Backpropagation
    # Output layer gradients
    dy_pred = -(2 / n) * (y - y_pred)
    dv1 = np.sum(dy_pred * h1)
    dv2 = np.sum(dy_pred * h2)
    dc = np.sum(dy_pred)

    # Gradients for hidden layer
    dh1 = dy_pred * v1 * relu_derivative(z1)
    dh2 = dy_pred * v2 * relu_derivative(z2)

    dw1 = np.sum(dh1 * x)
    db1 = np.sum(dh1)
    dw2 = np.sum(dh2 * x)
    db2 = np.sum(dh2)

    # Update parameters
    v1 -= learning_rate * dv1
    v2 -= learning_rate * dv2
    c -= learning_rate * dc
    w1 -= learning_rate * dw1
    b1 -= learning_rate * db1
    w2 -= learning_rate * dw2
    b2 -= learning_rate * db2

    # Print loss every 100 iterations
    if epoch % 100 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.4f}")
```

```python
# Final output
print(f"Final parameters: v1 = {v1:.4f}, v2 = {v2:.4f}, w1 = {w1:.4f}, w2 = {w2:.4f}, b1 = {b1:.4f}, b2 = {b2:.4f}, c = {c:.4f}")
```

The estimated function in terms of weights and biases is:

y_{\text{pred}} = v_1 \cdot \max(0, w_1 x + b_1) + v_2 \cdot \max(0, w_2 x + b_2) + c

Where:
  - w_1, w_2: Weights for the hidden layer neurons
  - b_1, b_2: Biases for the hidden layer neurons
  - v_1, v_2: Weights for the connections from the hidden layer to the output
  - c: Bias for the output neuron
  - \max(0, z): ReLU activation function
